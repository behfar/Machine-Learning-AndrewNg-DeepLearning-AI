```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
from utils import *

# data

data_x = np.array([1,2,3,4,5])
data_y = np.array([2,4,6,8,10])

#--------formulas--------

def pred(x,w,b):
    pred = w*x+b
    print("pred_Y:",pred)
    return pred

#-----+-+-+-+-+-+-+-+---

def error(x,y,w,b):
    j_wb = sum((y-(w*x+b))**2)
    print("j(w,b):",j_wb)
    return j_wb

#-----------------------
def derivative_w(x,y,w,b):
    derivative_w = sum(-2*x*(y-(w*x+b)))
    print("derivative_w:",derivative_w)
    return derivative_w

def derivative_b(x,y,w,b):
    derivative_b = sum(-2*(y-(w*x+b)))
    print("derivative_b:",derivative_b)
    return derivative_b

#------------------------

f = -0.01
l = 0.01

#--------------train_wb----------
def train_wb(w=0, b=0, number=0.01):

    while True:

        dw = derivative_w(data_x, data_y, w, b)
        db = derivative_b(data_x, data_y, w, b)

        if (f < dw< l) and  (f < db < l):
            break

        w -= number * dw
        b -= number * db

        print("w:", w, " \nb:", b)

    print("Final W:", w)
    print("Final B:", b)

    return w,b

#----------------------------------------------------

W,B=train_wb()

error(data_x,data_y,W,B)
pred(data_x,W,B)
print("   ")
print("The End_:)")

#The End_:)
```

```python
plt.clf()
plt.scatter(data_x, data_y)
plt.plot(data_x, W*data_x + B)
plt.show()
```
