```python
from sklearn.datasets import fetch_california_housing
import numpy as np
import pandas as pd

#-------data--------
data = fetch_california_housing(as_frame=True)
df = data.frame

X = df.drop("MedHouseVal", axis=1).values
y = df["MedHouseVal"].values
#-------------------

#-----------------formula-----------------------
def pred(X,w,b):
    y_pred = np.dot(X,w) + b
    return y_pred

def error(X,w,b,y):
    j_wb = np.sum((y - pred(X,w,b))**2)
    return j_wb

def derivative_w(X, y, w, b):
    pred_y = np.dot(X, w) + b
    deri_w = -2 * np.dot(X.T, (y - pred_y))
    return deri_w

def derivative_b(X, y, w, b):
    pred_y = np.dot(X, w) + b
    dreri_b = -2 * np.sum(y - pred_y)
    return dreri_b
#--------------------------------------------------

f = -0.01
l = 0.01

def train_wb(w=0, b=0, number=0.01):

    while True:

        dw = derivative_w(X, y, w, b)
        db = derivative_b(X, y, w, b)

        # FIX: make stopping condition valid for vectors
        for i in range (3000):

            w -= number * dw
            b -= number * db

            print("w:", w, "\nb:", b)

        print("Final W:", w)
        print("Final B:", b)

        return w, b

W, B = train_wb(w=np.zeros(X.shape[1]), b=0)
error(X,y,W,B)
```
